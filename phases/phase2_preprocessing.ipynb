```python
import cv2
print(cv2.__version__)
import cv2
import numpy as np
import matplotlib.pyplot as plt

IMAGE_PATH = r"F:\resume project\path project\input\map_image.png"

# ── STEP 1: Load & Resize ────────────────────────────────
image_bgr = cv2.imread(IMAGE_PATH)
image_resized = cv2.resize(image_bgr, (800, 400))

# ── STEP 2: Denoise ──────────────────────────────────────
image_denoised = cv2.fastNlMeansDenoisingColored(image_resized, None, 10, 10, 7, 21)

# ── STEP 3: BGR → HSV ────────────────────────────────────
image_hsv = cv2.cvtColor(image_denoised, cv2.COLOR_BGR2HSV)
image_rgb = cv2.cvtColor(image_denoised, cv2.COLOR_BGR2RGB)

# ── DISPLAY ALL 3 STEPS ──────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].imshow(cv2.cvtColor(image_resized, cv2.COLOR_BGR2RGB))
axes[0].set_title("Step 1: Resized (800x400)")
axes[0].axis('off')

axes[1].imshow(image_rgb)
axes[1].set_title("Step 2: Denoised")
axes[1].axis('off')

axes[2].imshow(image_hsv)
axes[2].set_title("Step 3: HSV Converted")
axes[2].axis('off')

plt.suptitle("Phase 2 — Preprocessing", fontsize=14)
plt.tight_layout()
plt.show()

print(f"Original Size  : {image_bgr.shape}")
print(f"Resized Shape  : {image_resized.shape}")
print(f"HSV Shape      : {image_hsv.shape}")

# ── SAVE FOR NEXT PHASE ──────────────────────────────────
np.save(r"F:\resume project\path project\input\preprocessed.npy", image_hsv)
print("Preprocessed image saved!")
```
